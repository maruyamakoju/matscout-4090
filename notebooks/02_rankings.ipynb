# 02 — Rankings
Explore the ranked candidates produced by `matscout run-campaign`.

```python
import warnings; warnings.simplefilter('ignore')
import pandas as pd
df = pd.read_parquet('../outputs/candidates_ranked.parquet')
print(len(df), 'ranked candidates')
cols = ['reduced_formula', 'final_score', 'battery_score', 'semiconductor_score',
        'catalyst_score', 'solar_score', 'co2_capture_score', 'novelty_score',
        'uncertainty_score', 'ml_e_above_hull', 'predicted_bandgap_ev']
df[[c for c in cols if c in df.columns]].head(20)
```

```python
# top 10 per campaign (deduped by formula)
for camp, field in [('battery','battery_score'), ('semiconductor','semiconductor_score'),
                    ('catalyst','catalyst_score'), ('solar','solar_score'),
                    ('co2_capture','co2_capture_score')]:
    sub = df[df['tags'].apply(lambda ts: camp in ts)].drop_duplicates('reduced_formula')
    print(f'\n=== {camp} top 10 ===')
    print(sub.sort_values('final_score', ascending=False)[['reduced_formula','final_score',field,'novelty_score']].head(10).to_string(index=False))
```

```python
# stability (proxy) vs novelty, sized by final score
ax = df.plot.scatter(x='ml_e_above_hull', y='novelty_score',
                     s=df['final_score'].fillna(0) * 120, alpha=0.4, figsize=(8, 6))
ax.set_title('stability vs novelty (size ~ final score)')
```
