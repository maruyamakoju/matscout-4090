Verify the environment, ML potentials, and a single relaxation work end-to-end.

```python
import warnings; warnings.simplefilter('ignore')
import torch
print('torch', torch.__version__, 'cuda', torch.cuda.is_available(),
      torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'cpu')
from matscout.relaxation.chgnet_relaxer import CHGNetRelaxer
from matscout.relaxation.mace_relaxer import MACERelaxer
print('chgnet available:', CHGNetRelaxer().available())
print('mace available:', MACERelaxer().available())
```

```python
from matscout.config.loader import load_global_config, load_campaign_config
from matscout.data.mp_client import fetch_mp_seeds
g = load_global_config(); c = load_campaign_config('battery')
seeds = fetch_mp_seeds(c, g, limit=20)
print(len(seeds), 'seeds')
for s in seeds[:5]:
    print(f'{s.reduced_formula:10s} nsites={s.nsites} sg={s.spacegroup_number} tags={s.tags[:3]}')
```

```python
# relax one structure with CHGNet
relaxer = CHGNetRelaxer(device='auto', max_steps=100)
res = relaxer.relax(seeds[0].get_structure())
print('ok', res.ok, 'E/atom', res.energy_per_atom, 'maxF', res.max_force_ev_a,
      'converged', res.converged, 'dV%', res.volume_change_pct)
```
