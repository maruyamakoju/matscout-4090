# 01 — Candidate generation
Generate candidates for a campaign and inspect the chemical diversity.

```python
import warnings; warnings.simplefilter('ignore')
import pandas as pd
from matscout.config.loader import load_global_config, load_campaign_config
from matscout.data.mp_client import fetch_mp_seeds
from matscout.generation.generate import generate_candidates

g = load_global_config()
campaign = 'solar'
c = load_campaign_config(campaign)
seeds = fetch_mp_seeds(c, g, limit=50)
cands = generate_candidates(seeds, c, g, n=300)
print(len(cands), 'candidates,', len({r.reduced_formula for r in cands}), 'unique formulas')
```

```python
df = pd.DataFrame([{'formula': r.reduced_formula, 'source': r.source,
                    'nsites': r.nsites, 'sg': r.spacegroup_number,
                    'elements': '-'.join(r.elements)} for r in cands])
df['source'].value_counts()
```

```python
# element frequency among generated candidates
from collections import Counter
cnt = Counter(e for r in cands for e in r.elements)
pd.Series(dict(cnt.most_common(20))).plot.bar(figsize=(10, 3), title=f'{campaign}: element frequency')
```
